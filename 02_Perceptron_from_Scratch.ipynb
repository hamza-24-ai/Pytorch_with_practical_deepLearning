```python
import numpy as np
import pandas as pd
import torch
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
import torch.nn as nn
from sklearn.metrics import accuracy_score
```

# Data Import from Github

```python
data_url = "https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv"
df = pd.read_csv(data_url)

print("First five rows of the data ")
print(df.head())
```

# Preprocessing the data

```python
# Checking about the shape of data
print("Data shape")
print(df.shape)

# Checking if any value is null
print("\n Checking null values ")
print(df.isnull().sum())
```

```python
# Drop unnecessary columns
df = df.drop(columns = ["id", "Unnamed: 32"])
print("Updated data in dataflow table")
print(df.isnull().sum())
```

```python
# Separating the input and output data
X = df.drop(columns = ["diagnosis"])
Y = df["diagnosis"]

# Now checking the data is it is separted or not
print("Input data ")
print(X.info())
print("\nOutput Data ")
print(Y.info())
```

```python
# Split the data

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

# Scaler the input data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("Scaler the model Input data")
print(X_train)

# Lable encoding the output Data
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)

print("\n Label Encoder model Output Data")
print(y_train)
```

# numpy to Tensor

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)

print("Input Data Tensor")
print(X_train_tensor)
print("\n Output Data Tensor")
print(y_train_tensor)
```

# Building Neural Network Model

```python
class SimpleNN():

  def __init__(self, X):
    self.weights = torch.rand(X.shape[1], 1, dtype = torch.float64, requires_grad=True)
    self.bias = torch.zeros(1, dtype = torch.float64, requires_grad = True)

  # Now Wrtie a Neural Network

  def forward_propagation(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_calculation(self,y_pred,y):
    y = y.double()
    y = y.view(-1,1)
    criteria = nn.BCELoss()
    loss = criteria(y_pred,y)
    return loss
```

# Important Parameters

```python
learning_rate = 0.1
epochs = 25
```

# Model Pipeline

```python
model = SimpleNN(X_train_tensor)
model.weights
```

```python
# Define models loop

for ep in range(epochs):
  # Forward propagation
  y_pred = model.forward_propagation(X_train_tensor)
  # Calculate Loss Function
  loss = model.loss_calculation(y_pred, y_train_tensor)

  # Backward propgation
  loss.backward()

  # Update Weight
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # Zero Gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  print(f"Epoch No : {ep+1} , Loss : {loss}")


```

# Model Evaluation

```python
# Model Evaluate by forward propagation
with torch.no_grad():
  y_pre = model.forward_propagation(X_test_tensor)
  y_pre = (y_pre > 0.5).float()

acc = accuracy_score(y_pre, y_test_tensor)

print(f"Accuracy Score of our perceptron : {acc:.2f}")
```
