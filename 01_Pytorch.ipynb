# Arithmatic Operations on tensors

```python
x = torch.rand(size  =(2,2))
print(x)
```

```python
# Addition
a = x + 2
print(f"Addition Value \n {a}")
# multiplication
m = x * 100
print(f"Multiplication value \n {m}")
# division
d = m // 2
print(f"Division Value \n {d}")
# Power Value
p = x ** 2
print(f"Power Value \n {p}")

```

# CPU VS GPU

```python
import torch
import time

num = 10000

m1_c = torch.rand(size = (num,num))
m2_c = torch.rand(size = (num,num))

start_time = time.time()
mul_cpu = torch.matmul(m1_c,m2_c)
cpu_time = time.time() - start_time

print(f"Cpu Runing Time : {cpu_time:.4f}")

m1_g = m1_c.to("cuda")
m2_g = m2_c.to("cuda")

start = time.time()
result_gpu = torch.matmul(m1_g,m2_g)
torch.cuda.synchronize()
gpu_time = time.time() - start
print(f"GPU Runing Time : {gpu_time:.4f}")
```
